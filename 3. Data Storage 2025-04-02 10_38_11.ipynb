```
%pip install pyarrow
%pip install SQLAlchemy

# Import necessary libraries
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, DateType, IntegerType, DoubleType
import logging
import pandas as pd

# Setup logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Load processed data from CSV
#dataframe = pd.read_csv("/dbfs/tmp/processed_news.csv")
dataframe = spark.read.format("delta").load("/mnt/data/processed_news").toPandas()


if not dataframe.empty:
    spark = SparkSession.builder.appName("NewsDataProcessing").getOrCreate()
    
    schema = StructType([
        StructField("source", StringType(), True),
        StructField("title", StringType(), True),
        StructField("date_posted", DateType(), True),
        StructField("author", StringType(), True),
        StructField("url", StringType(), True),
        StructField("content", StringType(), True),
        StructField("word_count", IntegerType(), True),
        StructField("sentiment", StringType(), True),
        StructField("compound_score", DoubleType(), True)
    ])

    # Ensure the schema (database) exists
    spark.sql("CREATE DATABASE IF NOT EXISTS the_news")

    # Switch to the correct schema
    spark.sql("USE the_news")

    spark_df = spark.createDataFrame(dataframe, schema=schema)
    # Write data to Delta Lake table
    spark_df.write.format("delta").mode('overwrite').saveAsTable('news_table')

    logger.info("Data successfully written to Delta Lake table.")

    
    #spark_df.write.format("delta").mode('overwrite').saveAsTable('the_news.news_table')
    #logger.info("Data successfully written to Delta Lake table.")
```
